```python
from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModel
# tokenizer=AutoTokenizer.from_pretrained("bert-base-uncased")
model1=SentenceTransformer("all-MiniLM-L6-v2")
# model2=AutoModel.from_pretrained("bert-base-uncased")
model2 = SentenceTransformer(
    "sentence-transformers/all-mpnet-base-v2"
)
model3=SentenceTransformer("BAAI/bge-small-en-v1.5")
```

```python
import chromadb
client=chromadb.PersistentClient(path="content/chroma_db")
# collection=client.get_or_create_collection("documents")
docs = [
    """Generative AI is a branch of Artificial Intelligence that uses Large Language Models
    such as GPT, Gemini, Claude, and Llama to generate human-like text, code, images,
    summaries, translations, and business insights. Organizations use Generative AI
    for customer support, content creation, document processing, software development,
    and knowledge management systems.""",

    """Agentic AI extends Generative AI by enabling AI agents to autonomously plan,
    reason, make decisions, call tools, access APIs, retrieve information, execute
    tasks, and interact with external systems. Agentic AI frameworks such as LangGraph,
    CrewAI, AutoGen, and LangChain help build multi-agent systems capable of solving
    complex business workflows with minimal human intervention.""",

    """Retrieval-Augmented Generation (RAG) combines vector databases, embeddings,
    retrieval mechanisms, and Large Language Models to provide context-aware answers.
    Documents are converted into embeddings and stored in vector databases such as
    ChromaDB. When a user asks a question, relevant document chunks are retrieved and
    supplied to the LLM, reducing hallucinations and improving answer accuracy.""",

    """ChromaDB is an open-source vector database designed for Generative AI and Agentic AI
    applications. It stores embeddings generated from documents, PDFs, websites,
    and enterprise knowledge bases. ChromaDB performs semantic similarity search
    using vector representations, enabling intelligent retrieval for chatbots,
    AI assistants, and RAG systems.""",

    """Multi-Agent Systems consist of multiple AI agents working together to achieve
    a common goal. One agent may retrieve data, another may analyze it, another
    may generate reports, and another may communicate results. Such architectures
    are widely used in autonomous business operations, research assistants,
    software development automation, and enterprise workflow orchestration."""
]

# model1 - all-MiniLM-L6-v2
model1_collection = client.get_or_create_collection(
    name="all-MiniLM-L6-v2",
    metadata={"category": "all-MiniLM-L6-v2"}
)

model1_embeddings = model1.encode(docs).tolist()

model1_collection.add(
    ids=["m11", "m12", "m13", "m14", "m15"],
    embeddings=model1_embeddings,
    # metadatas=[{"source": "doc1"}, {"source": "doc2"}, {"source": "doc3"}, {"source": "doc4"}, {"source": "doc5"}
    documents=docs)

# model 2 - all-mpnet-base-v2

model2_collection = client.get_or_create_collection(
    name="all-mpnet-base-v2",
    metadata={"category": "all-mpnet-base-v2"}
)

model2_embeddings = model2.encode(docs).tolist()

model2_collection.add(
    ids=["m21", "m22", "m23", "m24", "m25"],
    embeddings=model2_embeddings,
    # metadatas=[{"source": "doc1"}, {"source": "doc2"}, {"source": "doc3"}, {"source": "doc4"}, {"source": "doc5"}
    documents=docs)

# model 3 - bge-small-en-v1.5

model3_collection = client.get_or_create_collection(
    name="bge-small-en-v1.5",
    metadata={"category": "bge-small-en-v1.5"}
)

model3_embeddings = model3.encode(docs).tolist()

model3_collection.add(
    ids=["m31", "m32", "m33", "m34", "m35"],
    embeddings=model3_embeddings,
    # metadatas=[{"source": "doc1"}, {"source": "doc2"}, {"source": "doc3"}, {"source": "doc4"}, {"source": "doc5"}
    documents=docs)
```

```python
for i in client.list_collections():
  print(i)
```

```python
client.delete_collection("documents")
```

```python
results = []
results.append(model1_collection.query(query_texts=["how does RAG works"],
                              n_results=3))
```

```python
query_embedding2 = model2.encode(
    ["how does RAG works"]
).tolist()

results.append(model2_collection.query(query_embeddings=query_embedding2,
                              n_results=3))
```

```python
query_embedding3 = model3.encode(
    ["how does RAG works"]
).tolist()

results.append(model3_collection.query(query_embeddings=query_embedding3,
                              n_results=3))
```

```python
for i in results:
  print(i['distances'][0])
```
